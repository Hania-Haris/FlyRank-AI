```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
declining_rate = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .mean() * 100
)

print(f"Pages marked as declining: {declining_rate:.2f}%")
```

```python
median_impressions = df["impressions_90d"].median()

print(f"Median 90-day impressions: {median_impressions:.1f}")
```

```python
median_ctr = df["ctr"].median()

print(f"Median CTR: {median_ctr:.3f}%")
```

```python
# Relationship between impressions and clicks
impressions_clicks_corr = (
    df[["impressions_90d", "clicks_90d"]]
    .corr()
    .loc["impressions_90d", "clicks_90d"]
)

# Relationship between content age and impressions
age_impressions_corr = (
    df[["content_age_days", "impressions_90d"]]
    .corr()
    .loc["content_age_days", "impressions_90d"]
)

# Relationship between average position and CTR
# Zero position means no position data, so exclude those rows.
position_ctr = df[
    ["avg_position", "ctr"]
].copy()

position_ctr = position_ctr[
    position_ctr["avg_position"] > 0
]

position_ctr_corr = (
    position_ctr
    .corr()
    .loc["avg_position", "ctr"]
)

print(f"Impressions vs clicks correlation: {impressions_clicks_corr:.3f}")
print(f"Content age vs impressions correlation: {age_impressions_corr:.3f}")
print(f"Average position vs CTR correlation: {position_ctr_corr:.3f}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*


At this stage, I can identify and measure associations between observable signals in the starter dataset. I can report correlations, grouped differences, and descriptive patterns, and I can investigate whether simple analytical or machine-learning methods provide additional evidence.

I cannot claim that a signal is a confirmed Google ranking factor, that one signal causes another to change, or that changing a particular content characteristic will definitely improve search performance. The available data is observational rather than the result of a controlled experiment.

For example, the correlation of 0.696 between impressions and clicks does not mean that increasing impressions will necessarily cause clicks to increase. Similarly, the near-zero correlation between content age and impressions does not prove that content age has no relationship with search performance; it only describes the simple linear association observed in this dataset.

I will therefore use careful language such as "associated with," "correlated with," "higher or lower on average," and "observed in this dataset." Any eventual recommendation will be treated as decision support rather than causal proof.
